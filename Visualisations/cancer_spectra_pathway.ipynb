```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Setup aesthetics
sns.set_theme(style="white", context="paper") 

# 2. Define Orders
evidence_order = ["Pathway-level", "Component-level", "Uncertain"]

plt.figure(figsize=(max(12, len(df["pathway"].unique())*0.6), 9))

# 3. Plotting
# Note: Ensure neglog10_p is numeric to allow the legend to scale correctly
ax = sns.scatterplot(
    data=df,
    x="pathway",
    y="cancer",
    size="- log10(p-adj)",
    hue="Evidence level",
    hue_order=evidence_order,
    sizes=(200, 800), 
    palette={
        "Pathway-level": "#076d6d",   
        "Component-level": "#08306b", 
        "Uncertain": "#91cf60"        
    },
    edgecolor="white",
    linewidth=0.8,
    alpha=0.85
)

# 4. Refined Aesthetics
ax.yaxis.grid(True, linestyle='--', which='major', color='grey', alpha=.2)
ax.xaxis.grid(True, linestyle='--', which='major', color='grey', alpha=.2)
ax.set_axisbelow(True)
sns.despine(left=True, bottom=True)

# 5. Labels & Ticks
ax.set_xlabel("Enriched Pathways", fontsize=25, fontweight='bold')
ax.set_ylabel("Cancer type", fontsize=25, fontweight='bold', labelpad=15)
plt.xticks(rotation=45, ha='right', fontsize=16)
plt.yticks(fontsize=16)

# 6. Fixing and Styling the Legend
# We use bbox_to_anchor to push the legend outside the plot area
leg = ax.legend(
    bbox_to_anchor=(1.02, 1), 
    loc='upper left', 
    borderaxespad=0,
    frameon=True,
    fontsize=14,          # Size of the labels (p-values and levels)
    title_fontsize=14,    # Size of the legend title
    labelspacing=1.2,     # Vertical spacing between items
    handletextpad=0.8     # Space between marker and text
)

leg.set_title("Evidence & Significance")

for handle in leg.legend_handles[1:4]:
    
    handle.set_markersize(15)

plt.tight_layout()
plt.savefig('bubble_fea_final.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
