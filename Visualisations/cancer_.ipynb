```python
import matplotlib.pyplot as plt

# Data for the charts
labels = ['Higher Capture', 'Lower Capture', 'Comparable Capture']
colors = ['#57ab56', '#d93e39', '#9b9b9b'] # Green, Red, Gray

# Percentages from the image
fastcore_values = [66.1, 7.1, 26.8]
swiftcore_values = [71.4, 14.3, 14.3]

# Create a figure with two subplots side-by-side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

# Plot 1: Sprintcore vs Fastcore
ax1.pie(fastcore_values, 
        labels=labels, 
        autopct='%1.1f%%', 
        startangle=90, 
        counterclock=False, 
        colors=colors,
        textprops={'fontsize': 12})
ax1.set_title('Sprintcore vs Fastcore', fontsize=16, fontweight='bold', pad=20)

# Plot 2: Sprintcore vs Swiftcore
ax2.pie(swiftcore_values, 
        labels=labels, 
        autopct='%1.1f%%', 
        startangle=90, 
        counterclock=False, 
        colors=colors,
        textprops={'fontsize': 12})
ax2.set_title('Sprintcore vs Swiftcore', fontsize=16, fontweight='bold', pad=20)

# Adjust layout to prevent overlapping
plt.tight_layout()

# Display the charts
plt.show()
```
